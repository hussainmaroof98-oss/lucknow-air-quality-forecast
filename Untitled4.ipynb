```python
small = 0     # counter for gaps of 6 hours or less
medium = 0    # counter for gaps of 7 to 24 hours
large = 0     # counter for gaps longer than 24 hours

for g in gap_lengths:        # take one gap length at a time and call it g
    if g <= 6:               # is this gap 6 hours or less?
        small = small + 1    # yes, so add 1 to the small counter
    elif g <= 24:            # if not, is it 24 hours or less?
        medium = medium + 1  # yes, so add 1 to the medium counter
    else:                    # otherwise it's longer than 24 hours
        large = large + 1    # add 1 to the large counter

print("small gaps (6h or less):", small)
print("medium gaps (7-24h):", medium)
print("large gaps (over 24h):", large)
```

```python

def fill_short_gaps(values, max_gap):
    values = list(values)       # make a copy of the data as a plain list
    n = len(values)             # n = how many values there are
    i = 0                       # i = the position we are looking at (start at 0)

    while i < n:                # keep going until we've looked at every position
        if pd.isnull(values[i]):            # is the value at position i blank?

            # we found the start of a gap, so now find where it ends
            j = i                           # j starts at the same place as i
            while j < n and pd.isnull(values[j]):   # keep moving right while still blank
                j = j + 1                   # move one position to the right

            gap_length = j - i              # number of blank hours in this gap
            has_before = i > 0              # is there a real value before the gap?
            has_after = j < n               # is there a real value after the gap?

            # fill only if the gap is short AND we have a value on both sides
            if gap_length <= max_gap and has_before and has_after:
                before = values[i - 1]                  # last real value before the gap
                after = values[j]                       # first real value after the gap
                step = (after - before) / (gap_length + 1)   # size of each equal step
                for k in range(gap_length):             # k goes 0, 1, 2, ... for each blank
                    values[i + k] = before + step * (k + 1)  # write the guessed value

            i = j                   # jump past the gap and continue from there
        else:
            i = i + 1               # value is not blank, so just move to the next position

    return values                   # give back the filled list
```

```python
def look_around(when_ist, hours=12):
    # turn the text time into a real Indian-time timestamp
    center = pd.Timestamp(when_ist).tz_localize("Asia/Kolkata")
    # work out the start and end of the window we want to see
    start = center - pd.Timedelta(hours=hours)
    end = center + pd.Timedelta(hours=hours)
    # print those rows
    print(df_full.loc[start:end, ["pm25", "pm10", "no2", "humidity","temperature"]])
```

```python
def build_timeline(raw):
    df = raw.copy()                                              # work on a copy, never on the original
    df["datetime_utc"] = pd.to_datetime(df["datetime_utc"], utc=True)   # text -> real dates
    df = df.sort_values("datetime_utc")                          # oldest first
    df = df.drop_duplicates("datetime_utc")                      # keep one row per timestamp
    df = df.set_index("datetime_utc")                            # use the date as the row label
    full_hours = pd.date_range(df.index.min(), df.index.max(), freq="h")   # every hour that should exist
    df = df.reindex(full_hours)                                  # missing hours become blank rows
    df.index.name = "datetime_utc"                               # keep the label name tidy
    return df
```

```python
def flag_faults(df):
    # start with every hour marked "fine" (False)
    bad = pd.Series(False, index=df.index)

    # rule 1: PM2.5 higher than PM10 is impossible
    rule1 = df["pm25"] > df["pm10"] + 20     # flag only if PM2.5 beats PM10 by more than 20     # True where PM2.5 is bigger than PM10
    bad = bad | rule1                   # "|" means OR: flag the hour if any rule says so
        # rule 2: a high value that barely changes for 4 hours in a row = frozen sensor
    window = df["pm25"].rolling(4)                              # look at 4 hours at a time
    barely_changes = (window.max() - window.min()) <= 2         # highest minus lowest is within 2
    is_high = window.min() >= 150                               # all 4 hours at or above 150
    frozen_end = barely_changes & is_high                       # True on the 4th hour of a frozen stretch

    rule2 = frozen_end.copy()
    for k in range(1, 4):                                       # also mark the 3 hours before it
        rule2 = rule2 | frozen_end.shift(-k, fill_value=False)
    bad = bad | rule2
        # rule 3: a high value far above the hours around it (a lone spike)
    around = pd.concat(
        [df["pm25"].shift(k) for k in [-6, -5, -4, -3, -2, -1, 1, 2, 3, 4, 5, 6]],
        axis=1,
    )                                                   # 12 columns: the 6 hours before and 6 after
    typical = around.median(axis=1)                     # the typical level around each hour
    enough_data = around.notna().sum(axis=1) >= 2       # at least 2 of the 12 must have a value
    rule3 = (df["pm25"] >= 150) & (df["pm25"] > 5 * typical) & enough_data
    bad = bad | rule3
    return bad                          # one True/False for every hour
```

```python
def stretch_flags(flags, hours=1):
    stretched = flags.copy()
    for k in range(1, hours + 1):                  # repeat for 1 hour, or more if we ask
        stretched = stretched | flags.shift(k, fill_value=False) | flags.shift(-k, fill_value=False)
    return stretched
```

```python
def clean_data(raw, known_faults=None, max_gap=6):
    # step 1: complete hourly timeline
    df = build_timeline(raw)

    # step 2: rule-based flags, widened by 1 hour on each side
    flags = flag_faults(df)
    flags = stretch_flags(flags, 1)

    # step 3: add the hand-checked fault windows, if any are given
    if known_faults is not None:
        for start_t, end_t in known_faults:
            flags.loc[start_t:end_t] = True

    # step 4: keep a record of the raw values we are about to blank
    report = df.loc[flags, ["pm25", "pm10"]].copy()

    # step 5: blank PM2.5 and PM10 in the flagged hours
    df.loc[flags, ["pm25", "pm10"]] = np.nan

    # step 6: fill short gaps (wind_direction is skipped on purpose)
    for col in ["pm25", "pm10", "no2", "o3", "temperature", "humidity", "wind_speed"]:
        df[col] = fill_short_gaps(df[col], max_gap)

    # step 7: Indian time and hour of day
    df["time_ist"] = df.index.tz_convert("Asia/Kolkata")
    df["hour_ist"] = df["time_ist"].dt.hour

    return df, report
```

```python
model_a = LinearRegression()
model_a.fit(X_train[["pm25"]], y_train)          # learn from the older 80%

pred_a = model_a.predict(X_test[["pm25"]])       # guess the newer 20%
print("Model A error on test rows:", round(mean_absolute_error(y_test, pred_a), 1))
print("multiplier:", round(model_a.coef_[0], 3), "| adjustment:", round(model_a.intercept_, 2))
```

```python
from sklearn.metrics import mean_squared_error, r2_score

guesses = {"simple guess": X_test["pm25"], "Model A": pred_a, "Model B": pred_b}
for name, g in guesses.items():
    mae = mean_absolute_error(y_test, g)
    rmse = np.sqrt(mean_squared_error(y_test, g))
    r2 = r2_score(y_test, g)
    bias = (g - y_test).mean()
    print(name, "| MAE", round(mae, 1), "| RMSE", round(rmse, 1), "| R2", round(r2, 2), "| bias", round(bias, 1))
```

```python
"""
Lucknow PM2.5 forecasting (Gomti Nagar station)

Steps: raw CSV -> clean_data -> add_features -> time-ordered split -> models.
Run in Google Colab with Google Drive already mounted.
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split


# ----------------------------------------------------------------------
# Settings
# ----------------------------------------------------------------------
BASE = "/content/drive/MyDrive/air-quality"
RAW_PATH = f"{BASE}/data/raw/gomti_nagar_hourly_raw.csv"
CLEAN_PATH = f"{BASE}/data/processed/gomti_nagar_hourly_clean.csv"
REPORT_PATH = f"{BASE}/data/processed/removed_hours_report.csv"

HORIZON = 6        # forecast this many hours ahead (1 = next hour)
MAX_GAP = 6        # gaps up to this many hours are filled, longer ones stay blank
SHOW_PLOTS = False

# Hand-checked sensor faults (UTC). Each pair is (first bad hour, last bad hour).
fault_windows = [
    ("2026-08-02 05:30", "2026-08-02 13:30"),   # PM2.5 of 683, other stations normal
    ("2026-09-25 06:30", "2026-09-25 06:30"),   # one lone 672 reading
    ("2025-05-26 14:30", "2025-05-26 23:30"),   # stuck near 384 for 10 hours
    ("2025-08-01 02:30", "2025-08-01 02:30"),   # one lone 478 reading
    ("2025-09-24 02:30", "2025-09-24 04:30"),   # 454, blank, 584
    ("2026-05-30 12:30", "2026-05-30 13:30"),   # 484 and 447, no other station saw it
    ("2025-09-23 18:30", "2025-09-23 19:30"),   # 112 and 251, not confirmed
    ("2025-10-20 16:30", "2025-10-20 22:30"),   # stuck at 286 on Diwali night
    ("2026-05-17 07:30", "2026-05-17 15:30"),   # spikes of about 310 every 3 hours
]

LAGS = [1, 3, 6, 24]
FEATURE_COLS = (
    ["hour_sin", "hour_cos", "month_sin", "month_cos", "dayofweek",
     "pm25", "pm10", "no2", "o3", "temperature", "humidity", "wind_speed"]
    + [f"pm25_lag{k}" for k in LAGS]
    + ["pm25_roll6", "pm25_roll24"]
)

GROUPS = {
    "history of pm25": ["pm25", "pm25_lag1", "pm25_lag3", "pm25_lag6", "pm25_lag24",
                        "pm25_roll6", "pm25_roll24"],
    "pm10": ["pm10"],
    "gases": ["no2", "o3"],
    "weather": ["temperature", "humidity", "wind_speed"],
    "time of day and year": ["hour_sin", "hour_cos", "month_sin", "month_cos", "dayofweek"],
}


# ----------------------------------------------------------------------
# Cleaning
# ----------------------------------------------------------------------
def fill_short_gaps(values, max_gap):
    """Fill blank stretches of max_gap hours or fewer with equal steps between
    the real value before the gap and the real value after it."""
    values = list(values)           # work on a plain list
    n = len(values)
    i = 0                           # position we are looking at

    while i < n:
        if pd.isnull(values[i]):    # start of a gap
            j = i
            while j < n and pd.isnull(values[j]):   # move right until the gap ends
                j = j + 1

            gap_length = j - i
            has_before = i > 0      # is there a real value before the gap?
            has_after = j < n       # is there a real value after the gap?

            if gap_length <= max_gap and has_before and has_after:
                before = values[i - 1]
                after = values[j]
                step = (after - before) / (gap_length + 1)
                for k in range(gap_length):
                    values[i + k] = before + step * (k + 1)

            i = j                   # jump past the gap
        else:
            i = i + 1

    return values


def build_timeline(raw):
    """Sort, drop repeated timestamps and insert a blank row for every missing hour."""
    df = raw.copy()
    df["datetime_utc"] = pd.to_datetime(df["datetime_utc"], utc=True)
    df = df.sort_values("datetime_utc")
    df = df.drop_duplicates("datetime_utc")
    df = df.set_index("datetime_utc")
    full_hours = pd.date_range(df.index.min(), df.index.max(), freq="h")
    df = df.reindex(full_hours)
    df.index.name = "datetime_utc"
    return df


def flag_faults(df):
    """Return True for every hour that looks like a sensor fault (PM2.5 only)."""
    bad = pd.Series(False, index=df.index)      # every hour starts as "fine"

    # rule 1: PM2.5 more than 20 above PM10 is impossible (PM2.5 is part of PM10)
    bad = bad | (df["pm25"] > df["pm10"] + 20)

    # rule 2: a high value that barely changes for 4 hours in a row = frozen sensor
    window = df["pm25"].rolling(4)
    barely_changes = (window.max() - window.min()) <= 2     # highest minus lowest within 2
    is_high = window.min() >= 150                           # all 4 hours at or above 150
    frozen_end = barely_changes & is_high                   # True on the 4th hour of the stretch
    rule2 = frozen_end.copy()
    for k in range(1, 4):                                   # also mark the 3 hours before it
        rule2 = rule2 | frozen_end.shift(-k, fill_value=False)
    bad = bad | rule2

    # rule 3: a high value far above the hours around it (a lone spike)
    around = pd.concat(
        [df["pm25"].shift(k) for k in [-6, -5, -4, -3, -2, -1, 1, 2, 3, 4, 5, 6]],
        axis=1,
    )                                                       # the 6 hours before and 6 after
    typical = around.median(axis=1)                         # typical level around each hour
    enough_data = around.notna().sum(axis=1) >= 2           # at least 2 of the 12 have a value
    rule3 = (df["pm25"] >= 150) & (df["pm25"] > 5 * typical) & enough_data
    bad = bad | rule3

    return bad


def stretch_flags(flags, hours=1):
    """Widen every flagged hour by `hours` on each side."""
    stretched = flags.copy()
    for k in range(1, hours + 1):
        stretched = stretched | flags.shift(k, fill_value=False) | flags.shift(-k, fill_value=False)
    return stretched


def clean_data(raw, known_faults=None, max_gap=MAX_GAP):
    """Full cleaning: timeline -> flag faults -> blank them -> fill short gaps -> IST columns.
    Returns the cleaned table and a report of the hours that were blanked."""
    df = build_timeline(raw)

    flags = stretch_flags(flag_faults(df), 1)
    if known_faults is not None:
        for start_t, end_t in known_faults:
            flags.loc[start_t:end_t] = True

    report = df.loc[flags, ["pm25", "pm10"]].copy()        # keep the raw values we remove
    df.loc[flags, ["pm25", "pm10"]] = np.nan

    # wind_direction is skipped on purpose (it is an angle)
    for col in ["pm25", "pm10", "no2", "o3", "temperature", "humidity", "wind_speed"]:
        df[col] = fill_short_gaps(df[col], max_gap)

    df["time_ist"] = df.index.tz_convert("Asia/Kolkata")
    df["hour_ist"] = df["time_ist"].dt.hour
    return df, report


# ----------------------------------------------------------------------
# Features and split
# ----------------------------------------------------------------------
def add_features(df, horizon=HORIZON):
    """Add time circles, lags, rolling averages and the target (PM2.5 `horizon` hours ahead)."""
    feat = df.copy()
    feat["month"] = feat["time_ist"].dt.month
    feat["dayofweek"] = feat["time_ist"].dt.dayofweek

    # hour and month as circles, so hour 23 is next to hour 0 and month 12 next to month 1
    feat["hour_sin"] = np.sin(2 * np.pi * feat["hour_ist"] / 24)
    feat["hour_cos"] = np.cos(2 * np.pi * feat["hour_ist"] / 24)
    feat["month_sin"] = np.sin(2 * np.pi * feat["month"] / 12)
    feat["month_cos"] = np.cos(2 * np.pi * feat["month"] / 12)

    for k in LAGS:
        feat[f"pm25_lag{k}"] = feat["pm25"].shift(k)            # PM2.5 k hours earlier
    feat["pm25_roll6"] = feat["pm25"].rolling(6, min_periods=4).mean()
    feat["pm25_roll24"] = feat["pm25"].rolling(24, min_periods=16).mean()

    feat["target"] = feat["pm25"].shift(-horizon)               # the answer we want to predict
    return feat


def make_split(feat):
    """Keep complete rows, then the older 80% for training and the newer 20% for testing."""
    data = feat[FEATURE_COLS + ["target"]].dropna()
    X, y = data[FEATURE_COLS], data["target"]
    return train_test_split(X, y, test_size=0.2, shuffle=False)    # never shuffle time data


def clip_to_training(X_other, X_train):
    """Cut unfamiliar values back to the range the model saw in training."""
    return X_other.clip(lower=X_train.min(), upper=X_train.max(), axis=1)


# ----------------------------------------------------------------------
# Models
# ----------------------------------------------------------------------
def scoreboard(y_true, guesses):
    """Print MAE, RMSE, R2 and bias for each set of guesses."""
    for name, g in guesses.items():
        mae = mean_absolute_error(y_true, g)
        rmse = np.sqrt(mean_squared_error(y_true, g))
        r2 = r2_score(y_true, g)
        bias = (g - y_true).mean()
        print(name, "| MAE", round(mae, 1), "| RMSE", round(rmse, 1),
              "| R2", round(r2, 2), "| bias", round(bias, 1))


def run_models(X_train, X_test, y_train, y_test):
    """Compare 'same as now' with Model A (pm25 only) and Model B (all inputs)."""
    X_test_clip = clip_to_training(X_test, X_train)

    model_a = LinearRegression().fit(X_train[["pm25"]], y_train)
    model_b = LinearRegression().fit(X_train, y_train)

    guesses = {
        "same as now": X_test["pm25"],
        "Model A": model_a.predict(X_test_clip[["pm25"]]),
        "Model B": model_b.predict(X_test_clip),
    }
    scoreboard(y_test, guesses)
    return model_a, model_b


def feature_group_experiment(X_train, y_train):
    """Practice test cut from the training rows: which input groups help?
    The real test set is not touched here."""
    cut = int(len(X_train) * 0.75)
    fit_X, val_X = X_train.iloc[:cut], X_train.iloc[cut:]
    fit_y, val_y = y_train.iloc[:cut], y_train.iloc[cut:]
    val_X_clip = clip_to_training(val_X, fit_X)

    def score(columns):
        m = LinearRegression().fit(fit_X[columns], fit_y)
        return mean_absolute_error(val_y, m.predict(val_X_clip[columns]))

    all_cols = list(X_train.columns)
    history = GROUPS["history of pm25"]

    print("same as now:", round((val_y - val_X["pm25"]).abs().mean(), 2))
    print("all", len(all_cols), "inputs:", round(score(all_cols), 2))

    for name, cols in GROUPS.items():                   # tear down: remove one group
        kept = [c for c in all_cols if c not in cols]
        print("without", name, ":", round(score(kept), 2))

    print("history only:", round(score(history), 2))    # build up: add one group
    for name, cols in GROUPS.items():
        if name != "history of pm25":
            print("history +", name, ":", round(score(history + cols), 2))

    print("history + time + weather:",
          round(score(history + GROUPS["time of day and year"] + GROUPS["weather"]), 2))
    print("all, but without pm10 and gases:",
          round(score([c for c in all_cols if c not in GROUPS["pm10"] + GROUPS["gases"]]), 2))


# ----------------------------------------------------------------------
# Optional plots
# ----------------------------------------------------------------------
def plot_eda(df_final):
    plt.figure(figsize=(14, 4))
    plt.plot(df_final.index, df_final["pm25"], linewidth=0.5)
    plt.title("PM2.5 at Gomti Nagar (cleaned)")
    plt.ylabel("PM2.5 (µg/m³)")
    plt.show()

    df_final.groupby("hour_ist")["pm25"].mean().plot(
        kind="bar", figsize=(10, 3), title="Average PM2.5 by hour of day (IST)")
    plt.show()


# ----------------------------------------------------------------------
# Run everything
# ----------------------------------------------------------------------
def main():
    raw = pd.read_csv(RAW_PATH)
    df_final, report = clean_data(raw, known_faults=fault_windows)

    print(df_final.shape)                                      # expect (14111, 10)
    print(len(report), "hours marked as faulty")
    print(df_final["pm25"].isnull().sum(), "pm25 blanks after filling")

    df_final.to_csv(CLEAN_PATH, index_label="datetime_utc")
    report.to_csv(REPORT_PATH)
    if SHOW_PLOTS:
        plot_eda(df_final)

    feat = add_features(df_final, HORIZON)
    X_train, X_test, y_train, y_test = make_split(feat)
    print(len(X_train), "training rows:", X_train.index.min(), "to", X_train.index.max())
    print(len(X_test), "test rows:    ", X_test.index.min(), "to", X_test.index.max())

    run_models(X_train, X_test, y_train, y_test)
    feature_group_experiment(X_train, y_train)


if __name__ == "__main__":
    main()
```
